```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

**Prediction for Circuit A:**

The qubit starts in the state |0⟩. Applying one X gate flips the state from |0⟩ to |1⟩. Therefore, I expect all 100 measurements to give the result 1 in an ideal simulator.


```python
# Circuit A: One X gate
circuit_A = QuantumCircuit(1, 1)

circuit_A.x(0)
circuit_A.measure(0, 0)

# Display the circuit
print(circuit_A.draw())

# Execute for 100 shots
simulator = AerSimulator()
result_A = simulator.run(circuit_A, shots=100).result()
counts_A = result_A.get_counts()

# Display results
print("Circuit A results:", counts_A)

# Display histogram
plot_histogram(counts_A, title="Circuit A: One X Gate")
plt.show()
```

**Prediction for Circuit B:**

The qubit starts in |0⟩. The first X gate changes it to |1⟩, and the second X gate changes it back to |0⟩. Therefore, I expect all 100 measurements to give 0 in an ideal simulator.


```python
# Circuit B: Two X gates
circuit_B = QuantumCircuit(1, 1)

circuit_B.x(0)
circuit_B.x(0)
circuit_B.measure(0, 0)

# Display the circuit
print(circuit_B.draw())

# Execute for 100 shots
result_B = simulator.run(circuit_B, shots=100).result()
counts_B = result_B.get_counts()

# Display results
print("Circuit B results:", counts_B)

# Display histogram
plot_histogram(counts_B, title="Circuit B: Two X Gates")
plt.show()
```

**Prediction for Circuit C:**

The Hadamard (H) gate transforms the initial state |0⟩ into an equal superposition of |0⟩ and |1⟩. Therefore, each measurement has a 50% probability of producing 0 and a 50% probability of producing 1. I expect approximately 50 zeros and 50 ones out of 100 shots, but the exact counts may differ.


```python
# Circuit C: One H gate
circuit_C = QuantumCircuit(1, 1)

circuit_C.h(0)
circuit_C.measure(0, 0)

# Display the circuit
print(circuit_C.draw())

# Execute for 100 shots
result_C = simulator.run(circuit_C, shots=100).result()
counts_C = result_C.get_counts()

# Display results
print("Circuit C results:", counts_C)

# Display histogram
plot_histogram(counts_C, title="Circuit C: One H Gate")
plt.show()
```

```python
print("Circuit A: One X gate")
display(circuit_A.draw(output="mpl"))

print("Circuit B: Two X gates")
display(circuit_B.draw(output="mpl"))

print("Circuit C: One H gate")
display(circuit_C.draw(output="mpl"))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

plot_histogram(counts_A, ax=axes[0])
axes[0].set_title("Circuit A: One X Gate")

plot_histogram(counts_B, ax=axes[1])
axes[1].set_title("Circuit B: Two X Gates")

plot_histogram(counts_C, ax=axes[2])
axes[2].set_title("Circuit C: One H Gate")

plt.tight_layout()
plt.show()
```

**Conclusion**

In this experiment, three single-qubit quantum circuits were created and executed using Qiskit with 100 shots per circuit.

Circuit A used one X gate and produced the measurement outcome 1, as predicted. Circuit B used two X gates, returning the qubit to its initial state |0⟩ and producing the outcome 0. Circuit C used an H gate, creating an equal superposition of the two basis states. Its measurements produced a probabilistic distribution of zeros and ones.

The experiment demonstrates that quantum gates affect measurement outcomes. The X gate flips a qubit, applying the X gate twice returns it to its original state, and the H gate creates a state with equal measurement probabilities. The histograms help compare theoretical predictions with experimental counts.
